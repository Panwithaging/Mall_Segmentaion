## Customer Segmentation using K-Means Clustering

In this project, we perform customer segmentation using the K-Means clustering algorithm. The objective is to group customers based on similar purchasing behavior using Age, Annual Income, and Spending Score.

Different clustering evaluation techniques such as the Elbow Method and Silhouette Score are used to determine the optimal number of clusters. Finally, Principal Component Analysis (PCA) is used to visualize the customer segments in two dimensions.

##About the Dataset

**Source:** https://www.kaggle.com/datasets/vjchoudhary7/customer-segmentation-tutorial-in-python/data

The dataset contains customer information collected from a shopping mall. The objective is to identify groups of customers with similar characteristics using unsupervised learning techniques.

### Dataset Columns

- **CustomerID** – Unique identifier assigned to each customer.
- **Gender** – Gender of the customer (Male/Female).
- **Age** – Age of the customer in years.
- **Annual Income (k$)** – Customer's annual income measured in thousands of dollars.
- **Spending Score (1–100)** – Score assigned by the mall based on customer spending behavior and purchasing habits.

## Feature Selection

Select the numerical features that best represent customer behavior.

The following features are used for clustering:

- Age
- Annual Income (k$)
- Spending Score (1–100)

Gender is excluded because the objective is to cluster customers based on behavioral and financial characteristics rather than demographic categories.

```python
x=data[["Age","Annual Income (k$)","Spending Score (1-100)"]]
```

```python
scaler=StandardScaler()
```

```python
x_scaled=scaler.fit_transform(x)
```

## Elbow Method

Determine the optimal number of clusters by plotting the Within Cluster Sum of Squares (WCSS) for different values of K.

The elbow point indicates where adding additional clusters results in only marginal improvement.

```python
#elbow method
inertia=[]
for i in range(1,11):
    Kmeans=KMeans(
        n_clusters=i,
        random_state=42,
        n_init=10
    )
    Kmeans.fit(x_scaled)
    inertia.append(Kmeans.inertia_)
```

```python
plt.figure(figsize=(8,5))

plt.plot(range(1,11),inertia,marker="o")
plt.title("Elbow Method")
plt.xlabel("No. of clusters")
plt.ylabel("inertia")
```

## Silhouette Score

Evaluate clustering quality using the Silhouette Score.

A higher silhouette score indicates that customers are well matched to their own cluster and well separated from neighboring clusters.

```python
scores=[]
for i in range(2,11):
    Kmeans=KMeans(
        n_clusters=i,
        random_state=42,
        n_init=10
    )
    labels=Kmeans.fit_predict(x_scaled)
    score=silhouette_score(x_scaled,labels)
    scores.append(score)
    print(f"k{i} silhouette_score :{score:.3f}")
```

## Train Final K-Means Model

Based on the Elbow Method and Silhouette Score, the optimal number of clusters was selected as **K = 6**.

The final K-Means model is trained using these six customer segments.

```python
Kmeans=KMeans(n_clusters=6,
random_state=42,
n_init=10
)
cluster=Kmeans.fit_predict(x_scaled)
```

```python
cluster.shape
```

```python
data["clusters"]=cluster
```

## Cluster Summary

Calculate the average Age, Annual Income, and Spending Score for each cluster to understand the characteristics of different customer groups.

```python
cluster_summary=data.groupby("clusters").mean(numeric_only=True)
```

Visualize the customer clusters using the first two principal components. Each color represents a different customer segment identified by the K-Means algorithm.
